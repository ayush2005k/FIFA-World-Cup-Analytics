```python
# ==========================================================
# CELL 191 — FINAL ANALYSIS SETUP
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

pd.set_option("display.max_columns", None)

print("Final analysis environment loaded.")
```

```python
# ==========================================================
# CELL 192 — PROJECT PATHS
# ==========================================================

PROJECT_ROOT = Path("..")

DATA_DIR = PROJECT_ROOT / "data"
INTERIM_DIR = DATA_DIR / "interim"
PROCESSED_DIR = DATA_DIR / "processed"

print("Project root:")
print(PROJECT_ROOT)

print("\nProcessed directory:")
print(PROCESSED_DIR)
```

```python
# ==========================================================
# CELL 193 — LOAD PLAYER DATA
# ==========================================================

possible_files = [
    INTERIM_DIR / "players_clean.csv",
    INTERIM_DIR / "players_cleaned.csv",
    INTERIM_DIR / "players.csv",
    PROCESSED_DIR / "players_clean.csv",
    PROCESSED_DIR / "players_cleaned.csv",
    PROCESSED_DIR / "players.csv"
]

player_file = None

for file in possible_files:
    if file.exists():
        player_file = file
        break

if player_file is None:
    raise FileNotFoundError(
        "Player dataset not found."
    )

players_final = pd.read_csv(player_file)

print("Loaded:")
print(player_file)

print("\nShape:")
print(players_final.shape)

display(players_final.head())
```

```python
# ==========================================================
# CELL 194 — LOAD TEAM DATA
# ==========================================================

possible_team_files = [
    INTERIM_DIR / "teams_clean.csv",
    INTERIM_DIR / "teams.csv",
    PROCESSED_DIR / "teams_clean.csv",
    PROCESSED_DIR / "teams.csv"
]

team_file = None

for file in possible_team_files:
    if file.exists():
        team_file = file
        break

if team_file is None:
    raise FileNotFoundError(
        "Team dataset not found."
    )

teams_final = pd.read_csv(team_file)

print("Loaded:")
print(team_file)

print("\nShape:")
print(teams_final.shape)

display(teams_final.head())
```

```python
# ==========================================================
# CELL 195 — LOAD MATCH DATA
# ==========================================================

possible_match_files = [
    INTERIM_DIR / "matches_clean.csv",
    INTERIM_DIR / "matches.csv",
    PROCESSED_DIR / "matches_clean.csv",
    PROCESSED_DIR / "matches.csv"
]

match_file = None

for file in possible_match_files:
    if file.exists():
        match_file = file
        break

if match_file is None:
    raise FileNotFoundError(
        "Match dataset not found."
    )

matches_final = pd.read_csv(match_file)

print("Loaded:")
print(match_file)

print("\nShape:")
print(matches_final.shape)

display(matches_final.head())
```

```python
# ==========================================================
# CELL 196 — TOURNAMENT OVERVIEW
# ==========================================================

print("=" * 70)
print("FIFA WORLD CUP 2026 — DATASET OVERVIEW")
print("=" * 70)

print(
    "Teams:",
    teams_final["team"].nunique()
    if "team" in teams_final.columns
    else len(teams_final)
)

print(
    "Players:",
    players_final["player"].nunique()
    if "player" in players_final.columns
    else len(players_final)
)

print(
    "Matches:",
    len(matches_final)
)

print("\nPlayer positions:")

if "position" in players_final.columns:
    display(
        players_final["position"]
        .value_counts()
        .to_frame("players")
    )
```

```python
# ==========================================================
# CELL 197 — TOP SCORERS
# ==========================================================

if "goals" in players_final.columns:

    top_scorers = (
        players_final[
            [
                "player",
                "team",
                "position",
                "goals",
                "minutes"
            ]
        ]
        .copy()
        .sort_values(
            "goals",
            ascending=False
        )
        .head(15)
    )

    print("=" * 70)
    print("TOP SCORERS")
    print("=" * 70)

    display(top_scorers)

else:

    print("Goals column not available.")
```

```python
# ==========================================================
# CELL 198 — TOP ASSIST PROVIDERS
# ==========================================================

if "assists" in players_final.columns:

    top_assists = (
        players_final[
            [
                "player",
                "team",
                "position",
                "assists",
                "minutes"
            ]
        ]
        .copy()
        .sort_values(
            "assists",
            ascending=False
        )
        .head(15)
    )

    print("=" * 70)
    print("TOP ASSIST PROVIDERS")
    print("=" * 70)

    display(top_assists)

else:

    print("Assists column not available.")
```

```python
# ==========================================================
# CELL 199 — TEAM GOAL ANALYSIS
# ==========================================================

home_goals = (
    matches_final
    .groupby("home_team")["home_score"]
    .sum()
    .rename("goals_for_home")
)

away_goals = (
    matches_final
    .groupby("away_team")["away_score"]
    .sum()
    .rename("goals_for_away")
)

team_goals = pd.concat(
    [
        home_goals,
        away_goals
    ],
    axis=1
).fillna(0)

team_goals["total_goals"] = (
    team_goals["goals_for_home"]
    +
    team_goals["goals_for_away"]
)

team_goals = (
    team_goals
    .sort_values(
        "total_goals",
        ascending=False
    )
    .reset_index()
    .rename(
        columns={
            "index": "team"
        }
    )
)

print("=" * 70)
print("TEAM GOAL RANKING")
print("=" * 70)

display(team_goals.head(15))
```

```python
# ==========================================================
# CELL 200 — TEAM GOAL VISUALIZATION
# ==========================================================

top_teams = team_goals.head(15)

plt.figure(figsize=(10, 7))

plt.barh(
    top_teams["team"][::-1],
    top_teams["total_goals"][::-1]
)

plt.xlabel("Goals")
plt.ylabel("Team")

plt.title(
    "Top Teams by Goals Scored"
)

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# CELL 201 — PLAYER PERFORMANCE BY POSITION
# ==========================================================

position_summary = (
    players_final
    .groupby("position")
    .agg(
        players=("player", "count"),
        total_goals=("goals", "sum"),
        total_assists=("assists", "sum"),
        avg_goals=("goals", "mean"),
        avg_assists=("assists", "mean"),
        avg_minutes=("minutes", "mean")
    )
    .reset_index()
)

position_summary = position_summary.sort_values(
    "total_goals",
    ascending=False
)

print("=" * 70)
print("PLAYER PERFORMANCE BY POSITION")
print("=" * 70)

display(position_summary)
```

```python
# ==========================================================
# CELL 202 — TOP ATTACKING PLAYERS
# ==========================================================

attacking_positions = [
    "FW",
    "MF",
    "FW,MF",
    "MF,FW"
]

attackers = players_final[
    players_final["position"].isin(attacking_positions)
].copy()

top_attackers = (
    attackers[
        [
            "player",
            "team",
            "position",
            "goals",
            "assists",
            "goals_per90",
            "assists_per90",
            "minutes"
        ]
    ]
    .sort_values(
        ["goals", "assists"],
        ascending=False
    )
    .head(20)
)

print("=" * 70)
print("TOP ATTACKING PLAYERS")
print("=" * 70)

display(top_attackers)
```

```python
# ==========================================================
# CELL 203 — DEFENSIVE LEADERS
# ==========================================================

defensive_positions = [
    "DF",
    "DF,MF",
    "MF,DF",
    "DF,FW"
]

defenders = players_final[
    players_final["position"].isin(defensive_positions)
].copy()

defenders["defensive_actions"] = (
    defenders["tackles_won"].fillna(0)
    +
    defenders["interceptions"].fillna(0)
)

top_defenders = (
    defenders[
        [
            "player",
            "team",
            "position",
            "tackles_won",
            "interceptions",
            "defensive_actions",
            "fouls",
            "minutes"
        ]
    ]
    .sort_values(
        "defensive_actions",
        ascending=False
    )
    .head(20)
)

print("=" * 70)
print("DEFENSIVE LEADERS")
print("=" * 70)

display(top_defenders)
```

```python
# ==========================================================
# CELL 204 — GOALKEEPER LEADERS
# ==========================================================

goalkeepers_final = players_final[
    players_final["position"] == "GK"
].copy()

goalkeeper_columns = [
    "player",
    "team",
    "gk_games",
    "gk_minutes",
    "gk_saves",
    "gk_save_pct",
    "gk_clean_sheets",
    "gk_clean_sheets_pct"
]

available_gk_columns = [
    column
    for column in goalkeeper_columns
    if column in goalkeepers_final.columns
]

top_goalkeepers = (
    goalkeepers_final[
        available_gk_columns
    ]
    .sort_values(
        "gk_save_pct",
        ascending=False,
        na_position="last"
    )
)

print("=" * 70)
print("GOALKEEPER PERFORMANCE")
print("=" * 70)

display(top_goalkeepers.head(15))
```

```python
# ==========================================================
# CELL 205 — MINUTES VS GOALS
# ==========================================================

player_minutes_analysis = players_final[
    [
        "player",
        "team",
        "position",
        "minutes",
        "goals",
        "assists",
        "goals_per90"
    ]
].copy()

player_minutes_analysis = (
    player_minutes_analysis
    .dropna(subset=["minutes"])
)

print("=" * 70)
print("MINUTES VS GOALS")
print("=" * 70)

display(
    player_minutes_analysis
    .sort_values(
        "minutes",
        ascending=False
    )
    .head(20)
)
```

```python
# ==========================================================
# CELL 206 — MINUTES VS GOALS VISUALIZATION
# ==========================================================

plot_data = player_minutes_analysis[
    player_minutes_analysis["minutes"] > 0
].copy()

plt.figure(figsize=(10, 7))

plt.scatter(
    plot_data["minutes"],
    plot_data["goals"]
)

plt.xlabel("Minutes Played")
plt.ylabel("Goals")

plt.title(
    "Player Minutes vs Goals"
)

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# CELL 207 — HOME VS AWAY PERFORMANCE
# ==========================================================

home_wins = (
    matches_final["home_score"]
    >
    matches_final["away_score"]
).sum()

away_wins = (
    matches_final["away_score"]
    >
    matches_final["home_score"]
).sum()

draws = (
    matches_final["home_score"]
    ==
    matches_final["away_score"]
).sum()

home_away_summary = pd.DataFrame({
    "Result": [
        "Home Wins",
        "Draws",
        "Away Wins"
    ],
    
    "Matches": [
        home_wins,
        draws,
        away_wins
    ]
})

home_away_summary["Percentage"] = (
    home_away_summary["Matches"]
    /
    len(matches_final)
    * 100
).round(2)

print("=" * 70)
print("HOME VS AWAY PERFORMANCE")
print("=" * 70)

display(home_away_summary)
```

```python
# ==========================================================
# CELL 208 — MATCH SCORING PATTERNS
# ==========================================================

scoring_analysis = matches_final[
    [
        "home_team",
        "away_team",
        "home_score",
        "away_score"
    ]
].copy()

scoring_analysis["total_goals"] = (
    scoring_analysis["home_score"]
    +
    scoring_analysis["away_score"]
)

total_goals = scoring_analysis["total_goals"].sum()

average_goals = (
    scoring_analysis["total_goals"].mean()
)

highest_scoring_match = (
    scoring_analysis
    .sort_values(
        "total_goals",
        ascending=False
    )
    .head(10)
)

print("=" * 70)
print("MATCH SCORING ANALYSIS")
print("=" * 70)

print(
    f"Total goals: {int(total_goals)}"
)

print(
    f"Average goals per match: {average_goals:.2f}"
)

print("\nHighest-scoring matches:")

display(highest_scoring_match)
```

```python
# ==========================================================
# CELL 209 — TOURNAMENT STATISTICS SUMMARY
# ==========================================================

tournament_summary = pd.DataFrame({
    "Metric": [
        "Teams",
        "Players",
        "Matches",
        "Total Goals",
        "Average Goals per Match",
        "Home Wins",
        "Draws",
        "Away Wins"
    ],
    
    "Value": [
        teams_final["team"].nunique()
        if "team" in teams_final.columns
        else len(teams_final),

        players_final["player"].nunique()
        if "player" in players_final.columns
        else len(players_final),

        len(matches_final),

        int(total_goals),

        round(average_goals, 2),

        int(home_wins),

        int(draws),

        int(away_wins)
    ]
})

print("=" * 70)
print("TOURNAMENT SUMMARY")
print("=" * 70)

display(tournament_summary)
```

```python
# ==========================================================
# CELL 210 — FINAL KEY FINDINGS
# ==========================================================

# ----------------------------------------------------------
# TOP SCORER
# ----------------------------------------------------------

top_scorer = (
    players_final
    .dropna(subset=["goals"])
    .sort_values(
        "goals",
        ascending=False
    )
    .iloc[0]
)


# ----------------------------------------------------------
# TOP ASSIST PROVIDER
# ----------------------------------------------------------

top_assist_player = (
    players_final
    .dropna(subset=["assists"])
    .sort_values(
        "assists",
        ascending=False
    )
    .iloc[0]
)


# ----------------------------------------------------------
# HIGHEST-SCORING TEAM
# ----------------------------------------------------------

top_scoring_team = (
    team_goals
    .sort_values(
        "total_goals",
        ascending=False
    )
    .iloc[0]
)


# ----------------------------------------------------------
# BEST ML MODEL
# ----------------------------------------------------------

# Since this is a NEW notebook, calculate the model name
# from the saved ML comparison file.

model_comparison_file = (
    PROCESSED_DIR / "model_comparison.csv"
)

if model_comparison_file.exists():

    saved_model_results = pd.read_csv(
        model_comparison_file
    )

    saved_model_results = (
        saved_model_results
        .dropna(subset=["Accuracy"])
        .sort_values(
            "Accuracy",
            ascending=False
        )
    )

    if len(saved_model_results) > 0:

        best_model_name = (
            saved_model_results.iloc[0]["Model"]
        )

    else:

        best_model_name = "Not available"

else:

    best_model_name = "Not available"


# ----------------------------------------------------------
# MOST COMMON MATCH OUTCOME
# ----------------------------------------------------------

most_common_outcome = (
    home_away_summary
    .loc[
        home_away_summary["Matches"].idxmax(),
        "Result"
    ]
)


# ----------------------------------------------------------
# FINAL FINDINGS TABLE
# ----------------------------------------------------------

key_findings = pd.DataFrame({

    "Finding": [
        "Top scorer",
        "Top assist provider",
        "Highest-scoring team",
        "Average goals per match",
        "Most common match outcome",
        "Best ML model"
    ],

    "Value": [

        f"{top_scorer['player']} "
        f"({top_scorer['team']})",

        f"{top_assist_player['player']} "
        f"({top_assist_player['team']})",

        f"{top_scoring_team['team']} "
        f"({int(top_scoring_team['total_goals'])} goals)",

        round(
            average_goals,
            2
        ),

        most_common_outcome,

        best_model_name
    ]
})


# ----------------------------------------------------------
# DISPLAY
# ----------------------------------------------------------

print("=" * 70)
print("FINAL KEY FINDINGS")
print("=" * 70)

display(key_findings)
```
