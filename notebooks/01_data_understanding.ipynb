```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries Loaded Successfully")
```

```python
# File Paths

PLAYERS_PATH = "../data/raw/kaggle/players.csv"
TEAMS_PATH = "../data/raw/kaggle/teams.csv"
MATCHES_PATH = "../data/raw/kaggle/matches.csv"

# Read CSV Files

players = pd.read_csv(PLAYERS_PATH)
teams = pd.read_csv(TEAMS_PATH)
matches = pd.read_csv(MATCHES_PATH)

print("Datasets Loaded Successfully")
```

```python
print("="*60)
print("PLAYERS DATASET")
print("="*60)
print(players.shape)

print()

print("="*60)
print("TEAMS DATASET")
print("="*60)
print(teams.shape)

print()

print("="*60)
print("MATCHES DATASET")
print("="*60)
print(matches.shape)
```

```python
print("PLAYERS")
display(players.head())

print("TEAMS")
display(teams.head())

print("MATCHES")
display(matches.head())
```

```python
print("PLAYERS")
display(players.tail())

print("TEAMS")
display(teams.tail())

print("MATCHES")
display(matches.tail())
```

```python
print("="*60)
print("PLAYERS COLUMNS")
print("="*60)

for column in players.columns:
    print(column)

print("\n")

print("="*60)
print("TEAMS COLUMNS")
print("="*60)

for column in teams.columns:
    print(column)

print("\n")

print("="*60)
print("MATCHES COLUMNS")
print("="*60)

for column in matches.columns:
    print(column)
```

```python
print("="*60)
print("PLAYERS INFO")
print("="*60)

players.info()

print("\n")

print("="*60)
print("TEAMS INFO")
print("="*60)

teams.info()

print("\n")

print("="*60)
print("MATCHES INFO")
print("="*60)

matches.info()
```

```python
print("PLAYERS")
display(players.dtypes)

print("TEAMS")
display(teams.dtypes)

print("MATCHES")
display(matches.dtypes)
```

```python
print("="*60)
print("PLAYERS MISSING VALUES")
print("="*60)

display(players.isnull().sum().sort_values(ascending=False))

print()

print("="*60)
print("TEAMS MISSING VALUES")
print("="*60)

display(teams.isnull().sum().sort_values(ascending=False))

print()

print("="*60)
print("MATCHES MISSING VALUES")
print("="*60)

display(matches.isnull().sum().sort_values(ascending=False))
```

```python
print("Players Duplicate Rows :", players.duplicated().sum())

print("Teams Duplicate Rows :", teams.duplicated().sum())

print("Matches Duplicate Rows :", matches.duplicated().sum())
```

```python
print("="*60)
print("PLAYERS")
print("="*60)

display(players.describe(include="all"))

print("="*60)
print("TEAMS")
print("="*60)

display(teams.describe(include="all"))

print("="*60)
print("MATCHES")
print("="*60)

display(matches.describe(include="all"))
```

```python
print("Players Memory Usage")

display(players.memory_usage(deep=True))

print()

print("Teams Memory Usage")

display(teams.memory_usage(deep=True))

print()

print("Matches Memory Usage")

display(matches.memory_usage(deep=True))
```

```python
summary = pd.DataFrame({
    "Dataset": ["Players", "Teams", "Matches"],
    "Rows": [
        players.shape[0],
        teams.shape[0],
        matches.shape[0]
    ],
    "Columns": [
        players.shape[1],
        teams.shape[1],
        matches.shape[1]
    ],
    "Duplicate Rows": [
        players.duplicated().sum(),
        teams.duplicated().sum(),
        matches.duplicated().sum()
    ]
})

summary
```

```python
summary.to_csv(
    "../reports/dataset_summary.csv",
    index=False
)

print("Dataset Summary Saved Successfully")
```
